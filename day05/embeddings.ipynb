```python
from sentence_transformers import SentenceTransformer, util

# Doc's example sentences
sentences = [
    "A man is eating pasta.",
    "Someone is enjoying a bowl of spaghetti.",
    "The stock market dropped today.",
    "A woman plays the violin on stage.",
]

# Exercise: a small "held-out" test set — (sentence A, sentence B, should they be similar?)
pairs = [
    ("How do I reset my password?", "I forgot my login credentials.", True),
    ("The car won't start in the morning.", "My vehicle has engine trouble.", True),
    ("She adopted a puppy last week.", "They got a new dog recently.", True),
    ("How do I reset my password?", "The recipe needs two eggs.", False),
    ("The car won't start in the morning.", "Interest rates rose again.", False),
    ("It is very hot outside.", "It is very cold outside.", False),
]

for name in ["all-MiniLM-L6-v2", "BAAI/bge-small-en-v1.5", "sentence-transformers/all-mpnet-base-v2"]:
    model = SentenceTransformer(name)
    emb = model.encode(sentences, convert_to_tensor=True, normalize_embeddings=True)
    print(f"\n=== {name} ({model.get_sentence_embedding_dimension()} dims) ===")
    print(util.cos_sim(emb, emb).round(decimals=3))

    a = model.encode([p[0] for p in pairs], convert_to_tensor=True, normalize_embeddings=True)
    b = model.encode([p[1] for p in pairs], convert_to_tensor=True, normalize_embeddings=True)
    scores = [round(float(util.cos_sim(a[i], b[i])), 2) for i in range(len(pairs))]
    print("pair scores:", scores)
```

```python
from sentence_transformers import SentenceTransformer

my_pairs = [
    ("I got a new laptop.", "I got a new computer.", True),
    ("My phone won't turn on.", "My phone won't charge.", True),
    ("The person is sick.", "The person has a runny nose.", True),
    ("My car won't drive.", "My laptop won't turn on.", False),
    ("The bottle is broken.", "I don't have a charger.", False),
    ("It's going up.", "It's going down.", False),
]

for name in ["all-MiniLM-L6-v2", "BAAI/bge-small-en-v1.5", "sentence-transformers/all-mpnet-base-v2"]:
    model = SentenceTransformer(name)
    a = model.encode([p[0] for p in my_pairs], normalize_embeddings=True)
    b = model.encode([p[1] for p in my_pairs], normalize_embeddings=True)
    scores = (a * b).sum(axis=1)   # dot product = cosine, since normalized
    print(name, [round(float(s), 2) for s in scores])
```

**all-MiniLM-L6-v2** did best on my test set. It had the biggest gap between similar and unrelated pairs (0.35), compared with 0.33 for all-mpnet-base-v2 and 0.22 for bge-small-en-v1.5.

```python
import numpy as np, faiss, pandas as pd, time
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
corpus = pd.read_csv("../day02/data/raw/imdb_train.csv")["text"].head(20000).tolist()

emb = model.encode(corpus, batch_size=64, show_progress_bar=True, normalize_embeddings=True)
emb = np.array(emb, dtype="float32")
print(emb.shape)
```

```python
index = faiss.IndexFlatIP(emb.shape[1])   # exact search, inner product (= cosine)
index.add(emb)
print("Vectors in index:", index.ntotal)

query = model.encode(["a scary movie that kept me up at night"],
                     normalize_embeddings=True).astype("float32")
t = time.time()
D, I = index.search(query, k=5)
print(f"Search took {(time.time()-t)*1000:.1f} ms")

for rank, (score, idx) in enumerate(zip(D[0], I[0])):
    print(f"{rank+1}. score={score:.3f}  {corpus[idx][:120]}")
```

```python
# Save the index to a file
faiss.write_index(index, "imdb_minilm.faiss")

# Load it back, like you would after a restart
index2 = faiss.read_index("imdb_minilm.faiss")
print("Loaded vectors:", index2.ntotal)

# Check it gives the exact same search results as the original
D2, I2 = index2.search(query, k=5)
print("Same results as before:", (I2 == I).all())
```

```python
import pickle, faiss, time
from sentence_transformers import SentenceTransformer

index = faiss.read_index("wiki.faiss")
corpus = pickle.load(open("corpus.pkl", "rb"))
model = SentenceTransformer("all-MiniLM-L6-v2")

titles = sorted({c.split(":")[0] for c in corpus})
print(len(titles), "articles")
print(titles[:300])
```

```python
tests = [
    ("the fourth month of the year",                              "April"),
    ("month named after the Roman emperor Augustus",               "August"),
    ("painting, sculpture and other creative things people make",  "Art"),
    ("the mixture of gases that we breathe",                       "Air"),
    ("British mathematician who broke German codes in World War 2","Alan Turing"),
    ("Canadian singer who made the album Jagged Little Pill",      "Alanis Morissette"),
    ("computer program for drawing vector graphics",               "Adobe Illustrator"),
    ("spicy smoked sausage from France and Louisiana",             "Andouille"),
    ("growing crops and raising animals for food",                 "Farming"),
    ("adding, subtracting, multiplying and dividing numbers",      "Arithmetic"),
]

# Safety check: are all these articles actually in the corpus?
missing = [t for _, t in tests if not any(c.startswith(t + ":") for c in corpus)]
print("Titles not in corpus:", missing)

hits = 0
for query, title in tests:
    qv = model.encode([query], normalize_embeddings=True).astype("float32")
    t = time.perf_counter()
    D, I = index.search(qv, 5)
    ms = (time.perf_counter() - t) * 1000
    hit = any(corpus[i].startswith(title + ":") for i in I[0])
    hits += hit
    print(f"{'HIT ' if hit else 'MISS'}  {ms:5.1f} ms  {query}")

print(f"\nRecall@5 = {hits}/{len(tests)} = {hits/len(tests):.1f}")
```

```python
for query in ["British mathematician who broke German codes in World War 2",
              "adding, subtracting, multiplying and dividing numbers"]:
    qv = model.encode([query], normalize_embeddings=True).astype("float32")
    D, I = index.search(qv, 5)
    print(f"\nQUERY: {query}")
    for score, i in zip(D[0], I[0]):
        print(f"  {score:.3f}  {corpus[i][:100]}")
```
