```python
from transformers import pipeline
sa = pipeline("sentiment-analysis")
sa(["I loved this movie.", "Waste of two hours.", "It was... fine."])
```

```python
zsc = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
zsc("The Federal Reserve raised interest rates by 25 basis points.",
    candidate_labels=["economics", "sports", "technology", "entertainment"])
```

```python
ner = pipeline("ner", model="dslim/bert-base-NER", aggregation_strategy="simple")
ner("Barack Obama was born in Hawaii and graduated from Harvard.")
```

```python
qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
qa(question="Where did Obama graduate?",
   context="Barack Obama was born in Hawaii and graduated from Harvard.")
```

```python
summ = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")
article = ("The Federal Reserve raised interest rates by 25 basis points on Wednesday, "
           "its tenth increase in just over a year, as officials continued their effort "
           "to bring down inflation. Fed Chair Jerome Powell said the central bank would "
           "watch economic data closely before deciding on further increases. Markets "
           "reacted calmly, with major stock indexes ending the day slightly higher, "
           "while analysts said the Fed may be nearing the end of its rate-hiking cycle.")
summ(article, max_length=60, min_length=20)
```

Passing a list (batched input) returns a list with one result per text. Passing a single string returns the result for just that text.
